```python
import plotly.express as px
import pandas as pd
import numpy as np
```

```python
def ellipse_arc(x_center=0, y_center=0, a=1, b =1, start_angle=0, end_angle=2*np.pi, N=100, closed= False):
    t = np.linspace(start_angle, end_angle, N)
    x = x_center + a*np.cos(t)
    y = y_center + b*np.sin(t)
    path = f'M {x[0]}, {y[0]}'
    for k in range(1, len(t)):
        path += f'L{x[k]}, {y[k]}'
    if closed:
        path += ' Z'
    return path  
```

```python
mission = pd.DataFrame({'segment': [0, 1, 2, 4, 5, 7, 8, 10, 11, 12], 
                        'combat':[0, 0, 5000, 5000, 0, 0, 5000, 5000, 0, 0,]})


fig = px.line(mission, x='segment', y='combat',
        # template='simple_white', 
        width=1000, height=400,)
        #title="Strike")

x_start = 10.66
y_start=1700
fig.add_shape(type='line', x0=x_start, y0=y_start, x1=x_start+0.12, y1=y_start-600, line=dict(color="white", width=3), opacity=1)

fig.add_shape(type='line', x0=x_start, y0=y_start, x1=x_start+0.50, y1=y_start, line=dict(color="#1B3B7F", width=2), opacity=1)

fig.add_shape(type="path",
    path=ellipse_arc(x_center=x_start+0.50, y_center=y_start-150, a=0.25, b=150, start_angle=np.pi/2, end_angle=-np.pi/2, N=60),
    line_color="#1B3B7F",
    line=dict(color="#1B3B7F", width=2), fillcolor='white', opacity=1
)

fig.add_shape(type='line', x0=x_start-0.25, y0=y_start-300, x1=x_start+0.50, y1=y_start-300, line=dict(color="#1B3B7F", width=2), opacity=1)

fig.add_shape(type="path",
    path=ellipse_arc(x_center=x_start-0.25, y_center=y_start-300-150, a=0.25, b=150, start_angle=np.pi/2, end_angle=3*np.pi/2, N=60),
    line_color="#1B3B7F",
    line=dict(color="#1B3B7F", width=2), fillcolor='white', opacity=1
)

fig.add_shape(type='line', x0=x_start-0.25, y0=y_start-600, x1=x_start+0.12, y1=y_start-600, line=dict(color="#1B3B7F", width=2), opacity=1)

fig.update_traces(line_color="#1B3B7F")

fig.add_annotation(x=0, y=1, text="Catapult Takeoff<br>tropical day (89.8°F)<br>0 WOD", font=dict(family="Times New Roman", size=14),
ax=0, ay=-70)
fig.add_annotation(x=1.5, y=2500, text="Climb to cruise altitude", font=dict(family="Times New Roman", size=14),
                   ax=-70, ay=-50)
fig.add_annotation(x=3, y=5000, text="Combat 700 nm (outbound)", font=dict(family="Times New Roman", size=14),
                   ax=0, ay=-20)
fig.add_annotation(x=6, y=1, text="Mid-Mission<br>50 nm sea-level dashes<br>Mach ≥ 0.85<br>@ intermediate thrust", font=dict(family="Times New Roman", size=14),
                   ax=0, ay=-50 )
fig.add_annotation(x=9, y=5000, text="Combat 700 nm (inbound)", font=dict(family="Times New Roman", size=14),
                   ax=0, ay=-20)
fig.add_annotation(x=11.3, y=1700, text="20 min. loiter 10,000 ft<br>2 landing attempts", font=dict(family="Times New Roman", size=14),
                   ax=40, ay=-50)
fig.add_annotation(x=12, y=1, text="Arrestment Recovery", font=dict(family="Times New Roman", size=14),
                   ax=0, ay=-30 )
fig.update_xaxes(showticklabels=False, ticks="")
fig.update_yaxes(showticklabels=False, ticks="")
fig.update_layout(xaxis_title=None, yaxis_title=None, template='simple_white')
fig.write_image("report_figures/strike_mission.pdf")
```
