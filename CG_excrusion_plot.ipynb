```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import pandas as pd
import matplotlib.patheffects as patheffects

mpl.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "mathtext.fontset": "stix",
    "font.size": 16
})
```

```python
data = {'X_CG':[35.00439353, 35.23884546, 43.53062871, 43.12424536, 42.81164279, 43.57751909, 43.74945051, 43.38214248],
'Weight': [41056.0164, 41055.8877, 32881.33067, 32254.09759, 31783.64865, 29384.52637, 26894.25501, 26503.75859]}

pd.DataFrame(data)
```

```python
plt.figure(figsize=(11,6))
barbs = [patheffects.withTickedStroke()]
plt.plot(data['X_CG'], data['Weight'], marker='o', linestyle='-', linewidth=1, markerfacecolor='white', color="#1B3B7F",markersize=8)
#plt.scatter(data['X_CG'][0], data['Weight'][0], s=40, marker='o', color="#1B3B7F", zorder=50)
plt.axvline(x=28.11459141, color='tab:gray',path_effects=[patheffects.withTickedStroke(angle=135)])
plt.axvline(x=48.11459141, color='tab:gray', path_effects=[patheffects.withTickedStroke(angle=-45)])
plt.xlabel("CG Location, % MAC",)
plt.annotate("Forward CG Limit", xy=(27, 30000), rotation=90)
plt.annotate("Aft CG Limit", xy=(49, 30000), rotation=90)
plt.xlim([26, 50])
plt.annotate("takeoff", xy=(data['X_CG'][0]-2, data['Weight'][0]-200), fontsize=14)
plt.annotate("gear up", xy=(data['X_CG'][0]+1, data['Weight'][0]-200), fontsize=14)
plt.annotate("Forward Fuselage Tank", xy=(38, 35200), fontsize=14, rotation=-39)
plt.annotate("Fuselage Rear Tank 2", xy=(38, 32500), fontsize=14, rotation=0)
plt.annotate("Fuselage Rear Tank 3", xy=(37.5, 31500), fontsize=14, rotation=0)
plt.annotate("Drop 50 % Stores", xy=(39, 30000), fontsize=14, rotation=0)
plt.annotate("Wing Tank 25 %", xy=(39.8, 28000), fontsize=14, rotation=0)
plt.annotate("Gear Down Landing", xy=(38.5, 26300), fontsize=14, rotation=0)
plt.yticks([data['Weight'][0], data['Weight'][-1]], ["$W_o$", "$W_{land}$"])
plt.savefig("report_figures/CG_excrusion.pdf")
```
